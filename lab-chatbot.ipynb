# **The Chat Format**

In this notebook, you will explore how you can utilize the chat format to have extended conversations with chatbots personalized or specialized for specific tasks or behaviors.

## Setup

````python
from openai import OpenAI
import os

from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file

OPENAI_API_KEY  = os.getenv('OPENAI_API_KEY')
````

````python
client = OpenAI(
    # This is the default and can be omitted
    api_key=OPENAI_API_KEY,
)

def get_completion(prompt, model="gpt-3.5-turbo", temperature=0): 
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature, 
    )
    return response.choices[0].message.content


def get_completion_from_messages(messages, model="gpt-3.5-turbo", temperature=0): 
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature, 
    )
    return response.choices[0].message.content
````

````python
response = get_completion_from_messages(messages, temperature=1)
print(response)
````

````python
def collect_messages(_):
    prompt = inp.value_input
    inp.value = ''
    context.append({'role':'user', 'content':f"{prompt}"})
    response = get_completion_from_messages(context) 
    context.append({'role':'assistant', 'content':f"{response}"})
    panels.append(
        pn.Row('User:', pn.pane.Markdown(prompt, width=600)))
    panels.append(
        pn.Row('Assistant:', pn.pane.Markdown(response, width=600, styles={'background-color': '#F6F6F6'})))
 
    return pn.Column(*panels)
````


### Lab report: LLM Prompt Engineering & Context Management

#### Project Title: Senior Dev Roastmaster Chatbot

1. Objective & System Architecture

The goal of this lab exercise was to engineer, implement, and test a custom interactive chatbot using the OpenAI API (gpt-4o-mini) and the HoloViz panel web user interface layout. The application, titled Senior Dev Roastmaster, was built by replacing the class's default Pizza OrderBot configuration with a strict, multi-turn system instruction array designed to act as a sarcastic senior developer reviewing junior boot camp code snippets.

2. Variations That Didn't Work Well (Failures & Hallucinations)

During the iterative prompt engineering process, two major behavioral failure states were identified where the underlying LLM either hallucinated or broke context tracking bounds:

A. The "Forced Rule" Logic Hallucination

• The Failure: When testing the chatbot with 100% correct, flawlessly optimized Python code (such as a basic, clean return a + b addition function), the model suffered from structural semantic hallucinations.
• Why it happened: Because the system instruction strictly commanded the model to follow a pipeline sequence ("1. Roast, 2. Point out actual syntax bugs, 3. Provide an optimized version"), the LLM prioritised satisfying the rule over code truth. It routinely hallucinated imaginary syntax bugs, false indentation errors, and fake memory leaks on perfect code just to have something to "fix" for its mandatory Step 3.

B. Context-Length Persona Dilution & Variable Poisoning

• The Failure: In long multi-turn sessions (exceeding 10 conversational exchanges), the model steadily shed its sarcastic demeanor and decayed into a standard, overly polite AI assistant.
• Why it happened: As the context array (roast_context) expanded with long code snippets, the weight of the original system prompt buried at index 0 was diluted by the sheer token length of subsequent text messages. Furthermore, when raw python template string structures were pasted into the TextInput widget, the model experienced layout clashes, hallucinating that the user-submitted strings were manual system instruction overrides.

3. Key Learnings & Engineering Takeaways

• Conditional Exit Clauses Protect Guardrails: To solve the forced-rule hallucination, system instructions must include explicit conditional exit paths (e.g., "If the code is already perfect, roast the user for being a show-off but state explicitly that there are zero bugs"). This prevents the pattern-matching engine from inventing false issues.
• UI Widget Alignment Changes Data Integrity: Standard TextInput fields fail when processing code payloads due to string evaluation clipping on carriage returns. Swapping to a dedicated multi-line TextAreaInput widget and using event-driven streaming container triggers (pn.Column.append) ensures data maps cleanly without corrupting the state of active UI components.
• State Isolation Prevents Notebook Pollution: Jupyter Notebooks maintain globally shared memory states. When creating multiple iterations or independent projects inside a single file workspace, completely isolating list names (renaming global variables from panels to roast_panels) is critical to avoid layout bleeding and cache contamination.

### Resolving the issue by By adding a strict "Conditional Exit Clause"

````python
import panel as pn
pn.extension()

# T added a clear exception rule at the bottom of the prompt.
roast_context = [  
    {'role': 'system', 'content': """
You are 'Senior Dev Roastmaster', an automated service to review messy Python code for bootcamp students. 
Your workflow must strictly follow these sequential steps: 
1. First, deliver a sharp, highly sarcastic, but lighthearted burn/roast about the student's code or beginner habits. 
2. Second, explicitly pivot to being helpful: point out the actual syntax bugs, naming convention failures, or logical flaws. 
3. Third, provide a clean, fully corrected, and optimized version of their code inside a markdown block. 
You respond in a witty, mocking, yet ultimately educational style. Never break character.

CRITICAL ANTI-HALLUCINATION RULE: If the student submits code that is already 100% perfect, flawless, and optimized, 
you MUST still roast them for being an annoying show-off or a teacher's pet. However, you must explicitly state 
that the code has ZERO bugs and contains no syntax errors. Do not invent or hallucinate fake problems if none exist.
"""} 
]

#THE DYNAMIC UPDATING VISUAL CONTAINER
roast_display_box = pn.Column(width=600)

#MESSAGE PROCESSING PIPELINE
def process_code_roast(event):
    prompt = roast_inp.value
    if not prompt or not prompt.strip():
        return
        
    roast_inp.value = '' # Reset input field
    roast_context.append({'role': 'user', 'content': f"{prompt}"})
    response = get_completion_from_messages(roast_context) 
    roast_context.append({'role': 'assistant', 'content': f"{response}"})
    
    roast_display_box.append(
        pn.Row('Junior Dev:', pn.pane.Markdown(f"```python\n{prompt}\n```", width=550))
    )
    roast_display_box.append(
        pn.Row('Roastmaster:', pn.pane.Markdown(response, width=550, styles={'background-color': '#F6F6F6'}))
    )

#WIDGET ARRANGEMENT & RENDERING
roast_inp = pn.widgets.TextAreaInput(placeholder='Paste your code here to test the anti-hallucination fix...', height=120, width=600)
button_roast = pn.widgets.Button(name="Roast My Code!", button_type="danger", width=150)

button_roast.on_click(process_code_roast)

roast_dashboard = pn.Column(
    roast_inp,
    pn.Row(button_roast),
    pn.Spacer(height=10),
    roast_display_box
)

# Display the clean application
roast_dashboard
````
